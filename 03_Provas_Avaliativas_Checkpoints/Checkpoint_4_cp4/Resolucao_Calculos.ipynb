```python
import numpy as np
import matplotlib.pyplot as plt

estados = ['Online Direto', 'Cache Local (Offline-First)']
latencia = [450, 12]
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#130d24')
ax.set_facecolor('#130d24')
bars = ax.bar(estados, latencia, color=['#ef4444', '#10b981'], width=0.4)
ax.set_title('Tempo de Carregamento da Tela Inicial (ms)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_ylabel('Tempo de Resposta (ms)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='y', color='#34245c', alpha=0.5)
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2.0, yval + 10, f'{yval} ms', ha='center', va='bottom', color='#ffffff', fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
