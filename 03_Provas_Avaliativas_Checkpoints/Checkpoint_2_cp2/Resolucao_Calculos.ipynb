### Formulação Matemática Formal
$$\text{Presentation} \to \text{Domain (UseCases)} \leftarrow \text{Data (Repository Impl)}$$


### Passo 1: Camada de Domínio Pura (Domain Layer)
Entidades puras de negócio e Use Cases sem qualquer dependência do framework Android.


### Passo 2: Inversão de Controle com Interfaces de Repositório
A camada Domain define a interface; a camada Data implementa o repositório concreto.


### Passo 3: Mapeamento de Modelos (DTO para Entidade)
Mappers isolando contratos de rede externos da lógica central de negócio.


### Passo 4: Injeção de Dependências com Hilt / Koin
Provisão centralizada de instâncias Singleton de clientes HTTP e bancos locais.


### Passo 5: Conclusão Estrutural
100% dos Use Cases testáveis com testes unitários JUnit puros executados em microssegundos.


```python
import numpy as np
import matplotlib.pyplot as plt

camadas = ['Domain (Regra)', 'Data (Rede/BD)', 'Presentation (UI)']
cobertura = [98.5, 91.2, 82.4]
cores = ['#8b5cf6', '#06b6d4', '#a855f7']
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#130d24')
ax.set_facecolor('#130d24')
bars = ax.bar(camadas, cobertura, color=cores, width=0.45)
ax.set_ylim(0, 110)
ax.set_title('Cobertura de Testes Automatizados por Camada Clean (%)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_ylabel('Cobertura de Código (%)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='y', color='#34245c', alpha=0.5)
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2.0, yval + 2, f'{yval}%', ha='center', va='bottom', color='#ffffff', fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
