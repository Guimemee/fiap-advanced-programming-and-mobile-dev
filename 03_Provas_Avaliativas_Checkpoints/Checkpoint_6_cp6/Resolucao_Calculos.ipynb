### Passo 1: Análise Estática de Código (ktlint e detekt)
Bloqueio automático de Pull Requests que violem convenções de estilo e complexidade ciclomática.


### Passo 2: Execução de Testes Unitários de ViewModel
Testes assíncronos com TestDispatcher e Turbine para asserções de StateFlow.


### Passo 3: Build de Release com Minificação ProGuard / R8
Ofuscação de código e remoção de classes mortas reduzindo o APK de 48 MB para 14 MB.


### Passo 4: Assinatura Criptográfica do App Bundle
Assinatura automatizada com chaves seguras armazenadas em GitHub Secrets.


### Passo 5: Conclusão de Entrega
Deploy automatizado do aplicativo de monitoramento para a equipe de campo do Desafio Sanofi.


```python
import numpy as np
import matplotlib.pyplot as plt

versoes = ['Original (Debug)', 'Com R8 Otimizado (Release)']
tamanho_mb = [48.2, 14.1]
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#130d24')
ax.set_facecolor('#130d24')
bars = ax.bar(versoes, tamanho_mb, color=['#ef4444', '#8b5cf6'], width=0.4)
ax.set_title('Otimização de Tamanho do App Bundle (R8 ProGuard)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_ylabel('Tamanho do Arquivo (MB)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='y', color='#34245c', alpha=0.5)
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2.0, yval + 1, f'{yval} MB', ha='center', va='bottom', color='#ffffff', fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
