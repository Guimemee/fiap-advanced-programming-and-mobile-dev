### Formulação Matemática Formal
$$\text{UI State} = f(\text{ViewModel State}), \quad \text{StateFlow}\langle \text{ScreenUiState} \rangle$$


### Passo 1: Modelagem do Estado de UI Selado (Sealed Interface)
Definir estados Loading, Success(data) e Error(message).


### Passo 2: Gerenciamento de Estado no ViewModel
Exposição de _uiState como MutableStateFlow privado e uiState como StateFlow imutável.


### Passo 3: Consumo Reativo no Composable
Uso de collectAsStateWithLifecycle() para pausar emissões quando o app está em segundo plano.


### Passo 4: Sobrevivência à Destruição de Configuração
Retenção dos dados do ViewModel durante rotações de tela sem recarregamento de rede.


### Passo 5: Conclusão de Arquitetura
Interface fluida com recomposição granular apenas nos nós alterados da árvore Compose.


```python
import numpy as np
import matplotlib.pyplot as plt

frames = np.arange(1, 61)
fps_tradicional = 60 - np.random.exponential(1.5, 60)
fps_compose = np.full(60, 60.0) - np.random.uniform(0, 0.5, 60)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#130d24')
ax.set_facecolor('#130d24')
ax.plot(frames, fps_tradicional, 'r--', lw=1.5, label='Views Tradicionais XML (Quedas de Frames)')
ax.plot(frames, fps_compose, color='#8b5cf6', lw=2.5, label='Jetpack Compose (60 FPS Estável)')
ax.set_ylim(45, 65)
ax.set_title('Estabilidade de Taxa de Quadros (FPS) no Scroll', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Número do Frame Renderizado', color='#94a3b8')
ax.set_ylabel('Taxa de Quadros (FPS)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#34245c', alpha=0.5)
ax.legend(facecolor='#20163b', edgecolor='#8b5cf6', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
