### Formulação Matemática Formal
$$\text{suspend fun fetchTelemetry}(): \text{Response}\langle \text{TelemetryDto} \rangle$$


### Passo 1: Interface de Serviço Retrofit
Definição de endpoints com anotações @GET, @POST e suspensão nativa com Coroutines.


### Passo 2: Interceptor de Autenticação OkHttp
AuthInterceptor injetando header 'Authorization: Bearer <token>' automaticamente.


### Passo 3: Mecanismo de Renovação de Sessão (Authenticator)
Ao receber HTTP 401, dispara refresh token síncrono e repete a requisição original.


### Passo 4: Conversão de Erros de Rede em Resultados de Domínio
Empacotamento em monad Result.success() ou Result.failure() tipado.


### Passo 5: Conclusão de Resiliência
Retentativas exponenciais automáticas garantindo entrega sob redes móveis instáveis.


```python
import numpy as np
import matplotlib.pyplot as plt

tentativas = np.array([1, 2, 3, 4, 5])
atraso_s = 2**(tentativas - 1)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#130d24')
ax.set_facecolor('#130d24')
ax.plot(tentativas, atraso_s, 'o-', color='#06b6d4', lw=2.5, label='Backoff Exponencial (s)')
ax.set_title('Estratégia de Retentativa de Rede (Exponential Backoff)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Número da Retentativa', color='#94a3b8')
ax.set_ylabel('Tempo de Espera (segundos)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#34245c', alpha=0.5)
ax.legend(facecolor='#20163b', edgecolor='#06b6d4', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
